#### 01 - Evaluation - Data Preprocessing

In this notebook, we will convert input sentences in a txt file into CG3 readings format and export to another txt file for evaluation


```python
# Import cg3 and fst processing library 
import sys 
sys.path.append("../src")
import cg3_process as cg3 
```

```python
GRAMMAR_FILE = "../data/CG3_rules/Ojibwe.cg3"
INPUT_SENTENCE_FILE = "../data/evaluation/sample/ojibwe_sentences.txt"
FST_READINGS_OUTPUT = "../data/evaluation/sample/eval_original_fst_readings.txt"
FST_BINARY_FILE = "../data/fst/ojibwe.att"
```

```python
def read_file(filename:str) -> list[str]:
    """
    Reads a file and returns its content as a list of lines.
    
    :param filename: The name of the file to read.
    :return: A list of lines from the file.
    """
    try:
        with open(filename, 'r', encoding="utf-8") as file:
            return file.readlines()
    except Exception as e:
        print(f"Error reading file {filename}: {e}")
        return []
```

```python
sentence_list = read_file(INPUT_SENTENCE_FILE)

# quick check the sentence list
for i in range(5):
    print(sentence_list[i].strip())
```

```python
# Load cg3 library
cg3.is_cg3_available()
```

```python
fst = cg3.load_fst_parser(binary_file_path=FST_BINARY_FILE)
```

```python
# Test run the fst parser and convert to cg3 format 
sample_sentence = sentence_list[0].strip()
print(f"Sample sentence: {sample_sentence}")
input_readings = cg3.objiwe_sentence_to_cg3_format(ojibwe_sentence=sample_sentence, fst=fst)
print(f"Input readings: {input_readings}")
```

```python
output_lines = []
ending_punctuations = ".?!"
# Batch process the sentences
print("Running FST parser and converting to CG3 format...")
for i, sentence in enumerate(sentence_list):
    print(f"Processing sentence {i+1}/{len(sentence_list)}: {sentence.strip()}")
    
    # # check if the sentence ends with a period, if not so, add it
    # if not sentence.strip()[-1] in ending_punctuations:
    #     sentence = sentence.strip() + '.'
        
    # Convert Ojibwe sentence to CG3 format
    input_readings = cg3.objiwe_sentence_to_cg3_format(ojibwe_sentence=sentence.strip(), fst=fst)
    output_lines.append(input_readings) 
    
print("Completed")
```

```python
# Quick check the output lines
print(output_lines[0])
```

```python
def write_to_file(filename:str, lines:list[str]) -> None:
    """
    Writes a list of lines to a file.
    
    :param filename: The name of the file to write to.
    :param lines: A list of lines to write to the file.
    """
    try:
        with open(filename, 'w', encoding="utf-8") as file:
            for line in lines:
                file.write(line + "\n")
        print(f"Output written to {filename}")
    except Exception as e:
        print(f"Error writing to file {filename}: {e}")
```

```python
# Write the output to a file
write_to_file(FST_READINGS_OUTPUT, output_lines)
```
