## CG3 Example run: disambiguation

```python
import sys, os
sys.path.append(os.path.abspath(".."))
from fst_runtime.fst import Fst
from grammar_modules.fst import load_fst_parser
from grammar_modules.disambiguation import fst_parse_word, disambiguate, DISAMBIGUATION_PATH
from grammar_modules.dependency import parse_dependencies, cg3_to_conllu_block, DEPENDENCY_PATH
from treebank_modules.corpus import visualise_conllu
```

```python
fst = load_fst_parser()
```

```python
print(fst_parse_word("niin", fst))
fst_parse_word("niwaabamaa", fst)
```

```python
# english = Can you tap that tree over there?
sentence = "niin niwaabamaa."
```

```python
disambiguated_sent = disambiguate(sentence, DISAMBIGUATION_PATH, fst, verbose=True)
```

## CG3 Example run continued: dependency parsing

```python
deps = parse_dependencies(sentence, DEPENDENCY_PATH, DISAMBIGUATION_PATH, fst, verbose=True)
```

```python
conllu = cg3_to_conllu_block(deps, 1)
print(conllu)
```

```python
# write temp file for visualization
import tempfile
from pathlib import Path

with tempfile.NamedTemporaryFile(mode="w", suffix=".conllu", delete=False) as tmp:
    tmp.write(conllu)
    tmp_path = Path(tmp.name)

# visualize deps
visualise_conllu(tmp_path, sent_no=1, compact=True, collapse_punct=True)
tmp_path.unlink()
```
