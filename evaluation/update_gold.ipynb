# Updating Gold Standards

If the FST or CGs are updated, we will want to update the gold standards. In the case of disambiguation, if the FST is updated to include more analyses, we will update all gold standard parses affected by this. In the case of dependency parsing, if our CG supports new dependencies, we will want to look at all cases where the parses differ from the previous version.

## Updating disambig gold

To update the disambiguation gold standard, we first rewrite the FST system outputs (cell 1), then diff these with the previous system outputs (cell 2), to focus on manually updating cases where analyses are added/removed .

```python
from grammar_modules.disambiguation import REPO_ROOT
from eval_modules.data_io import read_opd_tsv, build_eval_sample, write_eval_artifacts


TSV_FILE = REPO_ROOT / "evaluation" / "eval_data" / "sample_disambig" / "sample_492.tsv"
FST_BINARY_PATH = REPO_ROOT / "data" / "fst" / "ojibwe20260711.fomabin"
OUTDIR = REPO_ROOT / "evaluation" / "eval_data" / "sample_disambig"

rows = read_opd_tsv(TSV_FILE, has_id_col=True)
write_eval_artifacts(rows, FST_BINARY_PATH, OUTDIR, "20260721_sample_492")
```

```python
from grammar_modules.disambiguation import REPO_ROOT
from eval_modules.diff import compare_cg3_readings, diff_table

OLD_FILE = REPO_ROOT / "evaluation" / "eval_data" / "sample_disambig" / "sample_492.txt"
NEW_FILE = REPO_ROOT / "evaluation" / "eval_data" / "sample_disambig" / "20260721_sample_492.txt"
CHANGED_BLOCKS_FILE = REPO_ROOT / "evaluation" / "eval_data" / "out" / "20260721_disambig_to_be_updated.txt"
DIFF_CSV_PATH = "eval_data/out/20260721_disambig_diff.csv"

diff_table(compare_cg3_readings(OLD_FILE, NEW_FILE, CHANGED_BLOCKS_FILE), DIFF_CSV_PATH)
```

## Updating dependency gold

To update the dependency parsing gold standard, we first recompute the system outputs on the disambiguation gold standard (cell 1 in [dependency_eval.ipynb](dependency_eval.ipynb)), and then diff with the old system outputs (cell 3 here) to find where dependencies have been added/changed (usually not removed, as the focus is on increasing coverage). 

```python
from grammar_modules.disambiguation import REPO_ROOT
from eval_modules.diff import compare_conllu_outputs, diff_table

OLD_FILE = REPO_ROOT / "evaluation" / "eval_data" / "sample_dep" / "opd_dep_492.conllu"
NEW_FILE = REPO_ROOT / "evaluation" / "eval_data" / "sample_dep" / "20260602_opd_dep_492.conllu"

diff_table(compare_conllu_outputs(OLD_FILE, NEW_FILE), "eval_data/out/20260602_dep_diff.csv")
```
