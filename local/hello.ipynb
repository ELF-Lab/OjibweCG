```python
import spacy
from spacy import displacy

words = [
    "ingii-pizindawaag",
    "niin",
    "ingiw",
    "chi-aya'aag",
    ]

morph = [
    "Verb.1SgSubj.3PlObj",
    "Pron.1Sg",
    "Dem.3Pl",
    "Noun.3Pl",
    ""
]

arcs = [
    {"start": 0, "end": 1, "label": "nsubj", "dir": "right"},
    {"start": 0, "end": 3, "label": "obj",   "dir": "right"},
    {"start": 2, "end": 3, "label": "det",   "dir": "right"},
]

doc_data = {
    "words": [
        {"text": w, "tag": m}
        for w, m in zip(words, morph)
    ],
    "arcs": arcs
}

svg = displacy.render(
    doc_data,
    style="dep",
    manual=True,
    jupyter=False,
    options={
        "distance": 130,
        "offset_x": 80,
        "compact": True
    }
)

with open("sent28_dep.svg", "w", encoding="utf-8") as f:
    f.write(svg)

print("Saved sent28_dep.svg")
```
