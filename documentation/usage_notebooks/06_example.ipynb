## CG3 Example run: disambiguation

```python
import sys, os
sys.path.append(os.path.abspath(".."))
from fst_runtime.fst import Fst
from disambiguation import load_fst_parser, fst_parse_word
from dependency import parse_dependencies, cg3_to_conllu_block
from corpus import visualise_conllu
```

```python
fst_binary_filename = "../data/fst/ojibwe.fomabin"
fst = load_fst_parser(fst_binary_filename)
cg3_disambiguation_file = "../data/rules/disambiguation.cg3"
cg3_dependency_file = "../data/rules/dependency.cg3"
```

```python
fst_parse_word("mitig", fst)
```

```python
# english = Can you tap that tree over there?
sentence = "Gidaa-ozhiga'waa na awedi mitig."
```

```python
from disambiguation import disambiguate
disambiguated_sent = disambiguate(sentence, cg3_disambiguation_file, fst, verbose=True)
```

## CG3 Example run continued: dependency parsing

```python
deps = parse_dependencies(sentence, cg3_dependency_file, cg3_disambiguation_file, fst, verbose=True)
```

```python
conllu = cg3_to_conllu_block(deps, 1)
print(conllu)
```

```python
# write temp file for visualization
import tempfile
from pathlib import Path

with tempfile.NamedTemporaryFile(mode="w", suffix=".conllu", delete=False) as tmp:
    tmp.write(conllu)
    tmp_path = Path(tmp.name)

# visualize deps
visualise_conllu(tmp_path, sent_no=1, compact=True, collapse_punct=True)
tmp_path.unlink()
```
